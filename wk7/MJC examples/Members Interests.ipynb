```python
import json
import collections
```

```python

with open('rmfi_commons.json') as json_data:
    d = json.load(json_data)
```

```python
for item in d['resources']:
    print(item['title'])
```

```python
for item in d['resources']:
    print(item['title'], item['description'])
```

```python
for i, item in enumerate(d['resources']):
    print(i, item['title'], item['description'])
```

```python
print(json.dumps(d['resources'][6], indent=4))
```

```python
visits_abroad = d['resources'][6]['data']
visits_abroad
```

```python
len(visits_abroad)
```

```python
mp_visits = collections.defaultdict(int)

for visit in visits_abroad:
    mp_visits[visit['member']] += 1

print(mp_visits)
```

```python
mp_counter = collections.Counter(mp_visits)
mp_counter.most_common(10)
```
