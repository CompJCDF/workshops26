```python
# lets create some data using Python dictionaries and lists
compj = {}

# add a key and value for the year
compj['year'] = '26/27'

# add a list of students
compj['students'] = []

# add some students to the list
compj['students'].append({'name': 'Asa'})
compj['students'].append({'name': 'Santiago'})
compj['students'].append({'name': 'Sayan'})
compj['students'].append({'name': 'Sethu'})
compj['students'].append({'name': 'Yara'})


# add somewhere to store some lecturers
compj['lecturers'] = []

# add some lecturers
compj['lecturers'].append({'name': 'Aidan'})
compj['lecturers'].append({'name': 'Martin'})

# output some data so we can see what it looks like
compj
```

```python
# write the data out as json
with open('compj.json', 'w') as outputfile:
    json.dump(compj, outputfile)

# go and open the file we just wrote in a text editor (or in jupyter) and see how the structure looks almost exactly like the above. 
# JSON is just a neat plain text way of storing data in structures like dictionaries and list
```

```python
# read the data in again, but store it in a new variable
with open('compj.json', 'r') as inputfile:
    cj = json.load(inputfile)
    print(cj)
```

```python
# anything we read in from a json file we can use just like normal python data structures. 
# We can access values using their keys, as the structure is just a dictionary, and we can loop through lists as normal
# this works for simple json structures like we have here, or for more complicated structures ...

print(cj['year'])

students = cj['students']
for s in students:
    print(s['name'].lower())
```

From the image, we can see that we have a `root` key that points to a dictionary with four other key/value pairs (`name`, `title`, `description`, `version`). We then have two more keys, `licenses` and `contributors` followed by our final, useful data structure with the key `resources`.

![Exapnding the Tree](jupyter-json-more.png)

Expanding this tree we can see the structure in the `resources` dictionary/object more clearly. We see there is only a single item in this list, which is an object/dictionary with multiple keys - one of which is `data` - the list of items that holds all the info we're interested in.

```python
# so really, all we need to deal with in this data structure is that `data` item
with open('SRMI.json', 'r') as inputfile:
    all_data = json.load(inputfile)

interesting_data = all_data['resources'][0]['data']
print(len(interesting_data))
```

```python
# we can loop through this list and do whatever we want to do with each data item. so for instance, we can filter by category:

category_5 = []

for d in interesting_data:
    if d['category_id'] == '5':
        category_5.append(d)

print(len(category_5))
```

```python
# defaultdict allows us to initialise a dictionary with a default value for all keys - so we don't need to check if a key already
# exists before we start interacting with its value
from collections import defaultdict

# setting up a defaultdict with an `int` as default value means that the dictionary will use a value of '0' for any keys
# it hasn't seen before
members = defaultdict(int)

for member in category_5:
    members[member['person_name']] += 1

print(members)
```
